```python
Libraries for data science :
1. NumPy = Used for numerical computations. It provides fast arrays, matrices, and mathematical functions.
2. Pandas = Used for data manipulation and analysis. It works with tables (DataFrames) to clean, filter, and organize data.
3. Matplotlib = Used for creating basic graphs and charts like line, bar, scatter, and pie charts.
4. Seaborn = A statistical data visualization library built on Matplotlib. It creates attractive and informative graphs.
5. Scikit-learn (sklearn) = Used for machine learning tasks such as classification, regression, clustering, and model evaluation.
6. SciPy = Used for scientific and mathematical computations like optimization, statistics, and linear algebra.
```

```python
#23. check for missing values =
print(df.isnull())
```

```python
#24. count missing values =
print(df.isnull().sum())
```

```python
#25.remove rows containing missing values =
df = df.dropna()

print(df)
```

```python
#26.replace missing values =
df.fillna(0, inplace=True)

print(df)
```

```python
#27. check for duplicate rows =
print(df.duplicated())
```

```python
#28. count duplicate rows =
print(df.duplicated().sum())
```

```python
#30.sort a DataFrame =
df = df.sort_values("year")

print(df)
```

```python
#31. filter rows =
print(df[df["year"] > 2020])
```

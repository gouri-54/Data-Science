```python
import pandas as pd

data = {
    "Name": ["Amit", "Neha", "Rahul", "Priya", "Karan",
             "Sneha", "Rohit", "Pooja", "Vikas", "Anjali"],
    "Age": [20, 21, 19, 22, 20, 21, 23, 20, 22, 19],
    "Percentage": [85, 90, 78, 88, 75, 92, 80, 86, 89, 91]
}

df = pd.DataFrame(data)

new_data = {
    "Name": ["Amit", "Neha", None, "Karan", "Rohit"],
    "Age": [20, None, 19, 20, 23],
    "Percentage": [85, 90, None, 75, 80]
}

new_df = pd.DataFrame(new_data)

df = pd.concat([df, new_df], ignore_index=True)

df["Remarks"] = ""

print(df)
```

```python
Q5. Write a Python program to get the number of observations, missing values, and duplicate values.
```

```python
import pandas as pd

data = {
    "Name": ["Amit", "Neha", "Rahul", "Priya", "Karan",
             "Sneha", "Rohit", "Pooja", "Vikas", "Anjali",
             "Amit", "Neha", None, "Karan", "Rohit"],
    "Age": [20, 21, 19, 22, 20, 21, 23, 20, 22, 19,
            20, None, 19, 20, 23],
    "Percentage": [85, 90, 78, 88, 75, 92, 80, 86, 89, 91,
                   85, 90, None, 75, 80],
    "Remarks": [""] * 15
}

df = pd.DataFrame(data)

print("Number of Observations :", len(df))

print("\nMissing Values")
print(df.isnull().sum())

print("\nDuplicate Rows :", df.duplicated().sum())
```

```python
Q6. Write a Python program to drop the 'Remarks' column from the DataFrame. Also drop all null and empty values. Print the modified data.
```

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Amit", "Neha", "Rahul", "Priya", "Karan",
             "Sneha", "Rohit", "Pooja", "Vikas", "Anjali",
             "Amit", "Neha", None, "Karan", "Rohit"],
    "Age": [20, 21, 19, 22, 20, 21, 23, 20, 22, 19,
            20, None, 19, 20, 23],
    "Percentage": [85, 90, 78, 88, 75, 92, 80, 86, 89, 91,
                   85, 90, None, 75, 80],
    "Remarks": [""] * 15
}

df = pd.DataFrame(data)

# Drop Remarks column
df = df.drop("Remarks", axis=1)

# Replace empty strings with NaN
df.replace("", np.nan, inplace=True)

# Remove rows containing null values
df = df.dropna()

print(df)
```

```python
Q7. Write a Python program to generate a line plot of Name vs Percentage
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Name": ["Amit", "Neha", "Rahul", "Priya", "Karan",
             "Sneha", "Rohit", "Pooja", "Vikas", "Anjali"],
    "Percentage": [85, 90, 78, 88, 75, 92, 80, 86, 89, 91]
}

df = pd.DataFrame(data)

plt.figure(figsize=(8,5))
plt.plot(df["Name"], df["Percentage"], marker="o")

plt.title("Name vs Percentage")
plt.xlabel("Name")
plt.ylabel("Percentage")

plt.grid(True)

plt.show()
```

```python
Note: If the dataset stores Height in centimeters, convert it to meters before calculating BMI.
```

```python
import pandas as pd

df = pd.read_csv("heights_weights.csv")

# Convert Height from cm to meters
height = df["Height"] / 100

# Calculate BMI
df["BMI"] = df["Weight"] / (height ** 2)

print(df.head())
```

```python
Q6. Write a Python program to find the maximum and minimum BMI.
```

```python
import pandas as pd

df = pd.read_csv("heights_weights.csv")

height = df["Height"] / 100

df["BMI"] = df["Weight"] / (height ** 2)

print("Maximum BMI :", df["BMI"].max())

print("Minimum BMI :", df["BMI"].min())
```
